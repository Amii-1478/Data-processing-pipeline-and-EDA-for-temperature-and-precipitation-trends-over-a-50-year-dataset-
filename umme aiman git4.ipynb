✅ Step 4: Create Dataset (50 Years)

```python
years = np.arange(1975, 2025)

# Simulated temperature (gradually increasing due to climate change)
temperature = 24 + 0.03*(years - 1975) + np.random.randn(50)*0.3

# Simulated rainfall (slightly fluctuating)
rainfall = 1000 + np.sin(np.linspace(0, 3*np.pi, 50))*100 + np.random.randn(50)*50

data = pd.DataFrame({
    "Year": years,
    "Temperature": temperature,
    "Rainfall": rainfall
})

print(data.head())
```

📈 Step 5: Temperature Trend Visualization

```python
plt.figure(figsize=(10,5))
plt.plot(data["Year"], data["Temperature"], color="red", marker="o")
plt.title("Temperature Trend Over 50 Years")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```

🌧️ Step 6: Rainfall Trend Visualization

```python
plt.figure(figsize=(10,5))
plt.plot(data["Year"], data["Rainfall"], color="blue", marker="o")
plt.title("Rainfall Trend Over 50 Years")
plt.xlabel("Year")
plt.ylabel("Rainfall (mm)")
plt.grid(True)
plt.show()
```

📊 Step 7: Combined Visualization


```python
plt.figure(figsize=(10,5))

plt.plot(data["Year"], data["Temperature"], label="Temperature (°C)", color="red")
plt.plot(data["Year"], data["Rainfall"]/50, label="Rainfall (scaled)", color="blue")

plt.title("Climate Change: Temperature vs Rainfall")
plt.xlabel("Year")
plt.ylabel("Values")
plt.legend()
plt.grid(True)
plt.show()
```
